```python
import networkx as nx
import matplotlib.pyplot as plt

def read_edges_from_file(file_path):
    edges = []
    with open(file_path, 'r') as file:
        for line in file:
            nodes = line.strip().split()
            if len(nodes) == 2:
                edges.append((int(nodes[0]), int(nodes[1])))
    return edges

file_path = 'F:/Aidin/Universities/Amirkabir (Master)/Lessons/Term 2/Analysis of Complex Networks/Assignments/HW2/Answer/Dataset/Weak-ties/first-graph.txt'
edges = read_edges_from_file(file_path)
G = nx.Graph()
G.add_edges_from(edges)
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G)
nx.draw(
    G,
    pos,
    with_labels=True,
    node_color='darkred',
    node_size=500,
    font_weight='bold',
    edge_color='gray'
)

plt.title("Graph from Edge List")
plt.show()
```
